```python
# imports

import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer

cancer = load_breast_cancer()

```

```python
cancer.keys()
```

```python
def answer_zero(dataset):
    # This function returns the number of feat
    # The assignment question description will
    return len(dataset['feature_names'])

print(answer_zero(cancer))
```

```python
def answer_one(dataset):
    df = pd.DataFrame(dataset['data'], columns=dataset['feature_names'])
    df['target'] = dataset['target']
    return df
    raise NotImplementedError()

print(answer_one(cancer))
```

```python
import pandas as pd

print(cancer.target_names)

def answer_two(df):
    counts = df['target'].value_counts()
    target = pd.Series(
        [counts[0], counts[1]],
        index=['malignant', 'benign'],
        name='target'
    )
    return target

df_cancer = answer_one(cancer)
answer_two(df_cancer)
```
